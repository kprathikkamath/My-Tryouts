```python
import os
import shutil
from pathlib import Path

print("=== Flattened Image Scraper & Exporter ===")

# Get paths via standard notebook inputs
input_folder_str = input("Enter the path to your INPUT folder: ").strip().strip('"').strip("'")
output_folder_str = input("Enter the path to your OUTPUT folder: ").strip().strip('"').strip("'")

input_path = Path(input_folder_str)
output_path = Path(output_folder_str)

# Validate input path
if not input_path.exists():
    print(f"\n❌ Error: The input folder path does not exist: {input_path}")
elif not input_path.is_dir():
    print(f"\n❌ Error: The input path is a file, not a folder: {input_path}")
else:
    # Create output directory if it doesn't exist
    output_path.mkdir(parents=True, exist_ok=True)
    
    # Supported image extensions
    image_extensions = {'.png', '.jpg', '.jpeg', '.webp', '.bmp', '.gif', '.tiff', '.ico'}
    
    copied_count = 0
    skipped_count = 0
    
    print(f"\nScanning all subfolders and flattening into: {output_path} ...\n")
    
    # rglob('*') recursively searches all nested folders
    for item in input_path.rglob('*'):
        if item.is_file():
            if item.suffix.lower() in image_extensions:
                dest_file = output_path / item.name
                
                # Handle filename collisions if duplicate names exist in different subfolders
                if dest_file.exists():
                    base = item.stem
                    ext = item.suffix
                    counter = 1
                    while dest_file.exists():
                        dest_file = output_path / f"{base}_{counter}{ext}"
                        counter += 1
                
                shutil.copy2(item, dest_file)
                print(f" [COPIED] {item.name} (from {item.parent.name})")
                copied_count += 1
            else:
                skipped_count += 1
                
    print(f"\n✨ Done! Successfully copied {copied_count} image(s) directly to: {output_path}")
    print(f"Ignored {skipped_count} non-image file(s).")
```
